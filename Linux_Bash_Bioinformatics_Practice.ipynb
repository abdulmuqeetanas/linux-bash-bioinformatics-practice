```python
%%bash
for base in A T G C; do
    count=$(grep -v "^>" TP53.fasta | tr -d '\n' | grep -o "$base" | wc -l)
    echo "$base: $count"
done
```

```python
%%bash

sequence=$(grep -v "^>" TP53.fasta | tr -d '\n')

length=${#sequence}

gc_count=$(echo -n "$sequence" | grep -o "[GC]" | wc -l)

awk -v gc="$gc_count" -v total="$length" 'BEGIN {
    printf "DNA Length: %d\n", total
    printf "GC Count: %d\n", gc
    printf "GC Content: %.2f%%\n", (gc / total) * 100
}'
```

```python
%%bash

sequence=$(grep -v "^>" TP53.fasta | tr -d '\n')
seq_length=${#sequence}
gc_count=$(printf "%s" "$sequence" | grep -o "[GC]" | wc -l)

awk -v total="$seq_length" -v gc="$gc_count" 'BEGIN {
    print "DNA Sequence Analysis Report"
    print "----------------------------"
    printf "Sequence Length: %d nucleotides\n", total
    printf "GC Count: %d\n", gc
    printf "GC Content: %.2f%%\n", (gc / total) * 100
}' > TP53_analysis_report.txt

echo "Report saved successfully!"
```
